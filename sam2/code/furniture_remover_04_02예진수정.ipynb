```python
# ============================================================
# 셀 3: 이미지 업로드 & 텍스트로 가구 선택
# ============================================================
%matplotlib inline
from google.colab import files
from IPython.display import display, clear_output
import ipywidgets as widgets
import io

print('📁 방 사진을 업로드하세요!')
uploaded = files.upload()

filename = list(uploaded.keys())[0]
image = Image.open(io.BytesIO(uploaded[filename])).convert('RGB')
image_np = np.array(image)

# 리사이즈
MAX_SIZE = 1024
h, w = image_np.shape[:2]
if max(h, w) > MAX_SIZE:
    scale = MAX_SIZE / max(h, w)
    image_np = cv2.resize(image_np, (int(w*scale), int(h*scale)))

h, w = image_np.shape[:2]
print(f'이미지 크기: W={w}, H={h}')

# 이미지 미리보기 (격자 포함)
fig, ax = plt.subplots(figsize=(14, 10))
ax.imshow(image_np)

# 격자 표시 (좌표 참고용)
grid_step = 100
for x in range(0, w, grid_step):
    ax.axvline(x, color='white', alpha=0.3, linewidth=0.5)
    ax.text(x, 10, str(x), color='white', fontsize=7, alpha=0.7)
for y in range(0, h, grid_step):
    ax.axhline(y, color='white', alpha=0.3, linewidth=0.5)
    ax.text(5, y, str(y), color='white', fontsize=7, alpha=0.7)

ax.set_title('📐 아래 입력창에 가구 이름과 위치를 입력하세요', fontsize=13)
plt.tight_layout()
plt.show()

# ── 가구 선택 UI ──────────────────────────────────────────
FURNITURE_PRESETS = {
    '소파':        {'ratio': (0.5, 0.6)},
    '침대':        {'ratio': (0.5, 0.5)},
    '책상':        {'ratio': (0.5, 0.4)},
    '의자':        {'ratio': (0.5, 0.5)},
    '테이블':      {'ratio': (0.5, 0.5)},
    '옷장':        {'ratio': (0.3, 0.4)},
    '서랍장':      {'ratio': (0.3, 0.5)},
    'TV 거치대':   {'ratio': (0.5, 0.6)},
    '책장':        {'ratio': (0.2, 0.4)},
    '기타 (직접입력)': {'ratio': (0.5, 0.5)},
}

style = {'description_width': '120px'}
layout = widgets.Layout(width='400px')

furniture_dropdown = widgets.Dropdown(
    options=list(FURNITURE_PRESETS.keys()),
    description='🪑 가구 종류:',
    style=style, layout=layout
)

custom_name = widgets.Text(
    placeholder='예: 행거, 화장대 ...',
    description='직접 입력:',
    style=style, layout=layout
)

pos_x = widgets.IntSlider(
    value=w//2, min=0, max=w-1, step=10,
    description='📍 X 위치:',
    style=style, layout=widgets.Layout(width='500px')
)
pos_y = widgets.IntSlider(
    value=h//2, min=0, max=h-1, step=10,
    description='📍 Y 위치:',
    style=style, layout=widgets.Layout(width='500px')
)

add_btn    = widgets.Button(description='➕ 추가', button_style='primary')
reset_btn  = widgets.Button(description='🗑️ 초기화', button_style='danger')
preview_btn = widgets.Button(description='👁️ 미리보기', button_style='info')
output_area = widgets.Output()

selected_points = []
selected_labels_text = []

def update_slider_from_preset(change):
    key = furniture_dropdown.value
    rx, ry = FURNITURE_PRESETS[key]['ratio']
    pos_x.value = int(w * rx)
    pos_y.value = int(ry * h)

furniture_dropdown.observe(update_slider_from_preset, names='value')

def on_add(b):
    key = furniture_dropdown.value
    name = custom_name.value.strip() if key == '기타 (직접입력)' and custom_name.value.strip() else key
    x, y = pos_x.value, pos_y.value
    selected_points.append([x, y])
    selected_labels_text.append(name)
    with output_area:
        clear_output(wait=True)
        print(f'✅ 추가됨: [{name}] → ({x}, {y})')
        for i, (pt, lb) in enumerate(zip(selected_points, selected_labels_text)):
            print(f'  {i+1}. {lb} → {pt}')

def on_reset(b):
    selected_points.clear()
    selected_labels_text.clear()
    with output_area:
        clear_output()
        print('🗑️ 초기화 완료')

def on_preview(b):
    with output_area:
        clear_output(wait=True)
        if not selected_points:
            print('⚠️ 추가된 가구가 없습니다!')
            return
        fig2, ax2 = plt.subplots(figsize=(12, 8))
        ax2.imshow(image_np)
        for (x, y), name in zip(selected_points, selected_labels_text):
            ax2.scatter(x, y, c='red', s=300, zorder=5, marker='*')
            ax2.text(x+12, y, name, color='red', fontsize=11,
                     bbox=dict(boxstyle='round,pad=0.2', fc='white', alpha=0.7))
        ax2.set_title(f'선택된 가구 {len(selected_points)}개', fontsize=13)
        ax2.axis('off')
        plt.tight_layout()
        plt.show()
        print('확인 후 셀 4 실행!')

add_btn.on_click(on_add)
reset_btn.on_click(on_reset)
preview_btn.on_click(on_preview)

# 초기 슬라이더 값 세팅
update_slider_from_preset(None)

display(widgets.VBox([
    widgets.HTML('<b>🪑 제거할 가구를 선택하세요</b>'),
    furniture_dropdown,
    custom_name,
    widgets.HTML('<b>📍 격자 이미지 보고 위치 조정</b>'),
    pos_x,
    pos_y,
    widgets.HBox([add_btn, preview_btn, reset_btn]),
    output_area,
]))
```

```python
# ============================================================
# 셀 5: LaMa 인페인팅 (가구 제거)
# ============================================================
import torch

def run_lama(image_np, mask_np):
    # 8의 배수로 패딩
    h, w = image_np.shape[:2]
    pad_h = (8 - h % 8) % 8
    pad_w = (8 - w % 8) % 8

    if pad_h > 0 or pad_w > 0:
        image_np = np.pad(image_np, ((0, pad_h), (0, pad_w), (0, 0)), mode='reflect')
        mask_np  = np.pad(mask_np,  ((0, pad_h), (0, pad_w)),          mode='reflect')

    img = image_np.astype(np.float32) / 255.0
    msk = (mask_np > 127).astype(np.float32)
    img_t = torch.from_numpy(img).permute(2, 0, 1).unsqueeze(0).cuda()
    msk_t = torch.from_numpy(msk).unsqueeze(0).unsqueeze(0).cuda()

    with torch.no_grad():
        batch  = {'image': img_t, 'mask': msk_t}
        batch  = lama_model(batch)
        result = batch['inpainted'][0].permute(1, 2, 0).cpu().numpy()
        result = (result * 255).clip(0, 255).astype(np.uint8)

    # 패딩 제거 (원본 크기로 복원)
    result = result[:h, :w]
    return result

print('🎨 LaMa 인페인팅 실행 중...')
result = run_lama(image_np, mask_np)

# Drive에 저장
OUTPUT_DIR = f'{DRIVE_DIR}/outputs'
os.makedirs(OUTPUT_DIR, exist_ok=True)

Image.fromarray(result).save(f'{OUTPUT_DIR}/empty_room.jpg')
Image.fromarray(image_np).save(f'{OUTPUT_DIR}/original.jpg')
Image.fromarray(mask_np).save(f'{OUTPUT_DIR}/mask.png')

# 비교 시각화
fig, axes = plt.subplots(1, 3, figsize=(20, 7))
axes[0].imshow(image_np)
axes[0].set_title('Original', fontsize=14)
axes[0].axis('off')
axes[1].imshow(mask_np, cmap='gray')
axes[1].set_title('Mask', fontsize=14)
axes[1].axis('off')
axes[2].imshow(result)
axes[2].set_title('Empty Room ✅', fontsize=14)
axes[2].axis('off')
plt.tight_layout()
plt.savefig(f'{OUTPUT_DIR}/comparison.png', dpi=150)
plt.show()

print(f'✅ 완료! Drive에 저장됨: {OUTPUT_DIR}')
print('결과가 맘에 들면 셀 6 실행!')
```

```python
# ============================================================
# 셀 8: 가구 정밀 추출 (배경 제거 개선)
# ============================================================
import os
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
import cv2
from collections import defaultdict

LIBRARY_DIR = f'{DRIVE_DIR}/furniture_library'
os.makedirs(LIBRARY_DIR, exist_ok=True)

def refine_mask(mask_np, edge_blur=3):
    """
    SAM2 마스크 정제:
    - dilate 제거 (배경 유입 방지)
    - 경계만 가우시안 블러로 부드럽게
    """
    # 작은 노이즈 제거 (erode → dilate)
    kernel_small = np.ones((3, 3), np.uint8)
    mask_clean = cv2.morphologyEx(mask_np, cv2.MORPH_OPEN, kernel_small)

    # 경계 부드럽게 (가우시안 블러를 알파값에만 적용)
    mask_float = mask_clean.astype(np.float32) / 255.0
    blur_size  = edge_blur * 2 + 1
    mask_blur  = cv2.GaussianBlur(mask_float, (blur_size, blur_size), 0)

    # 내부는 완전 불투명, 경계만 블렌딩
    inner = (mask_clean > 200).astype(np.float32)
    alpha = np.maximum(inner, mask_blur)

    return (alpha * 255).astype(np.uint8)

def extract_furniture(image_np, mask_np):
    """정제된 마스크로 가구만 RGBA 추출"""
    h_img, w_img = image_np.shape[:2]
    if mask_np.shape != (h_img, w_img):
        mask_np = cv2.resize(mask_np, (w_img, h_img), interpolation=cv2.INTER_NEAREST)

    binary = mask_np > 127
    if not np.any(binary):
        return None, None

    # RGBA 생성 (알파 = 정제된 마스크값 그대로)
    rgba = np.zeros((h_img, w_img, 4), dtype=np.uint8)
    rgba[..., :3] = image_np
    rgba[..., 3]  = mask_np   # 0~255 그라데이션 알파

    # bounding box (여백 최소화)
    rows = np.any(binary, axis=1)
    cols = np.any(binary, axis=0)
    rmin, rmax = np.where(rows)[0][[0, -1]]
    cmin, cmax = np.where(cols)[0][[0, -1]]

    pad  = 5   # 여백 최소화 (기존 10 → 5)
    rmin = max(0, rmin - pad)
    rmax = min(h_img, rmax + pad)
    cmin = max(0, cmin - pad)
    cmax = min(w_img, cmax + pad)

    cropped = rgba[rmin:rmax, cmin:cmax]
    return Image.fromarray(cropped), (rmin, rmax, cmin, cmax)

# ── 1. 포인트 그룹화 (핵심 수정!) ───────────────────────
if not selected_points:
    raise ValueError('⚠ 셀 3에서 가구를 먼저 선택하세요!')

# 레이블별로 포인트 좌표를 모읍니다.
label_to_points = defaultdict(list)
for pt, lbl in zip(selected_points, selected_labels_text):
    label_to_points[lbl].append(pt)

saved_furniture = []
print(f'총 {len(label_to_points)}종류의 가구 추출 시작...\n')

# ── 2. 그룹화된 단위로 SAM2 실행 ───────────────────────
for idx, (label, points) in enumerate(label_to_points.items()):
    print(f'[{idx+1}/{len(label_to_points)}] "{label}" 처리 중 (포인트 {len(points)}개 사용)...')

    predictor.set_image(image_np)

    # 🔥 여러 개의 포인트를 한 번에 모델에 전달합니다.
    masks, scores, _ = predictor.predict(
        point_coords=np.array(points),
        point_labels=np.array([1] * len(points)), # 모든 점을 포함(Positive)으로 설정
        multimask_output=True,
    )

    # 면적이 가장 큰 마스크 선택
    areas = [np.sum(mask) for mask in masks]
    best_mask_idx = np.argmax(areas)
    best_mask = masks[best_mask_idx].astype(bool)
    best_score = scores[best_mask_idx]

    h_img, w_img = image_np.shape[:2]
    if best_mask.shape != (h_img, w_img):
        best_mask = cv2.resize(best_mask.astype(np.uint8), (w_img, h_img), interpolation=cv2.INTER_NEAREST).astype(bool)

    # 마스크 정제 및 구멍 메우기
    mask_raw = (best_mask * 255).astype(np.uint8)
    close_kernel = np.ones((25, 25), np.uint8)
    mask_raw = cv2.morphologyEx(mask_raw, cv2.MORPH_CLOSE, close_kernel)
    contours, _ = cv2.findContours(mask_raw, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    cv2.drawContours(mask_raw, contours, -1, 255, thickness=cv2.FILLED)
    mask_refined = refine_mask(mask_raw, edge_blur=3)

    furniture_img, bbox = extract_furniture(image_np, mask_refined)
    if furniture_img is None: continue

    # 파일 저장 (이름이 같으면 덮어쓰거나 하나만 생성됨)
    name_safe = label.replace(' ', '_')
    save_path = f'{LIBRARY_DIR}/{name_safe}.png'
    furniture_img.save(save_path)

    saved_furniture.append({
        'name':  label,
        'file':  f'{name_safe}.png',
        'bbox':  bbox,
        'score': float(best_score),
    })
    print(f'  ✅ {name_safe}.png 저장 (score: {best_score:.3f})')

# ── 결과 시각화 ───────────────────────────────────────────
n = len(saved_furniture)
if n > 0:
    cols = min(n + 1, 4)
    rows = -(-( n + 1) // cols)   # ceiling division

    fig, axes = plt.subplots(rows, cols, figsize=(5 * cols, 5 * rows))
    if n == 0: # 에러 방지용 안전코드
        axes = np.array([axes])
    axes = np.array(axes).flatten()

    axes[0].imshow(image_np)
    for (x, y), name in zip(selected_points, selected_labels_text):
        axes[0].scatter(x, y, c='red', s=200, zorder=5, marker='*')
        axes[0].text(x+10, y, name, color='red', fontsize=9,
                     bbox=dict(boxstyle='round,pad=0.2', fc='white', alpha=0.7))
    axes[0].set_title('원본 + 선택 가구', fontsize=12)
    axes[0].axis('off')

    for i, info in enumerate(saved_furniture):
        ax = axes[i + 1]
        furn_img = Image.open(f'{LIBRARY_DIR}/{info["file"]}')
        fw, fh   = furn_img.size

        # 체커보드 배경
        checker = np.zeros((fh, fw, 3), dtype=np.uint8)
        for r in range(fh):
            for c in range(fw):
                checker[r, c] = [200,200,200] if (r//16 + c//16) % 2 == 0 else [255,255,255]
        fa    = np.array(furn_img)
        alpha = fa[..., 3:4] / 255.0
        prev  = (fa[..., :3] * alpha + checker * (1 - alpha)).astype(np.uint8)

        ax.imshow(prev)
        ax.set_title(f'{info["name"]}\nscore: {info["score"]:.3f}', fontsize=11)
        ax.axis('off')

    for j in range(n + 1, len(axes)):
        axes[j].axis('off')

    plt.suptitle(f'추출된 가구 {n}개', fontsize=14)
    plt.tight_layout()
    plt.show()

    print(f'\n📦 저장 완료:')
    for info in saved_furniture:
        print(f'  - {info["file"]} (score: {info["score"]:.3f})')
    print('\n셀 9에서 각 가구를 개별 배치하세요!')
```

```python
# ============================================================
# 셀 9: 재배치 UI (실시간 렌더링 + 완벽한 원근감 적용)
# ============================================================
from IPython.display import display, clear_output
import ipywidgets as widgets
import os
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
import cv2
import torch

# ── ZoeDepth 설치 및 로드 (최초 1회) ─────────────────────
if 'depth_model' not in dir():
    print('📦 ZoeDepth 설치 중...')
    os.system('pip install -q timm==0.6.13')
    os.system('pip install -q git+https://github.com/isl-org/ZoeDepth.git@v1.0')

    import shutil
    shutil.rmtree('/root/.cache/torch/hub', ignore_errors=True)
    cache_dir = '/root/.cache/torch/hub'
    for item in ['main.zip', 'master.zip', 'checkpoints/ZoeD_M12_NK.pt']:
        path = f'{cache_dir}/{item}'
        if os.path.exists(path):
            os.remove(path)

    ckpt_dir = f'{cache_dir}/checkpoints'
    ckpt_path = f'{ckpt_dir}/ZoeD_M12_NK.pt'
    os.makedirs(ckpt_dir, exist_ok=True)

    if not os.path.exists(ckpt_path):
        print('⬇️ ZoeDepth 체크포인트 다운로드 중...')
        os.system(
            f'wget -q --show-progress '
            f'https://github.com/isl-org/ZoeDepth/releases/download/v1.0/ZoeD_M12_NK.pt '
            f'-O {ckpt_path}'
        )

    print('🔄 ZoeDepth 모델 로드 중...')
    depth_model = torch.hub.load(
        'isl-org/ZoeDepth',
        'ZoeD_K',
        pretrained=True,
        trust_repo=True
    ).to('cuda').eval()
    print('✅ ZoeDepth 로드 완료!')

def get_depth_map(image_np):
    img_pil = Image.fromarray(image_np)
    with torch.no_grad():
        depth = depth_model.infer_pil(img_pil)
    depth_np = np.array(depth)
    d_min, d_max = depth_np.min(), depth_np.max()
    depth_norm = (depth_np - d_min) / (d_max - d_min + 1e-8)
    return depth_norm

def get_depth_at_point(depth_map, px, py):
    h, w = depth_map.shape
    px = np.clip(px, 0, w-1)
    py = np.clip(py, 0, h-1)
    x0, x1 = max(0, px-10), min(w, px+10)
    y0, y1 = max(0, py-10), min(h, py+10)
    return float(depth_map[y0:y1, x0:x1].mean())

def auto_scale_by_depth(furniture_img, depth_val, base_scale=1.0):
    depth_factor = 1.5 - depth_val
    return base_scale * depth_factor

# 🔥 핵심 수정 1: 진짜 3D 원근감을 시뮬레이션하는 완벽한 Warp 함수
def perspective_transform(img_pil, angle_h=0, angle_v=0):
    img_np = np.array(img_pil)
    h, w   = img_np.shape[:2]

    # 얼마나 찌그러뜨릴지 (최대 30%)
    warp_x = int(w * abs(angle_h) / 100)
    warp_y = int(h * abs(angle_v) / 100)

    # 3D 원근감을 위해, 멀어지는 쪽은 폭 뿐만 아니라 높이도 줄어들어야 합니다(사다리꼴)
    depth_h = int(h * abs(angle_h) / 200)
    depth_v = int(w * abs(angle_v) / 200)

    src = np.float32([[0, 0], [w, 0], [w, h], [0, h]])
    tl, tr, br, bl = [0, 0], [w, 0], [w, h], [0, h]

    # 수평 회전 (좌우) - angle_h가 양수면 오른쪽이 뒤로 멀어짐
    if angle_h > 0:
        tr = [w - warp_x, depth_h]
        br = [w - warp_x, h - depth_h]
    elif angle_h < 0:
        tl = [warp_x, depth_h]
        bl = [warp_x, h - depth_h]

    # 수직 회전 (상하) - angle_v가 양수면 위쪽이 뒤로 멀어짐
    if angle_v > 0:
        tl[0] += depth_v; tl[1] += warp_y
        tr[0] -= depth_v; tr[1] += warp_y
    elif angle_v < 0:
        bl[0] += depth_v; bl[1] -= warp_y
        br[0] -= depth_v; br[1] -= warp_y

    dst = np.float32([tl, tr, br, bl])

    M = cv2.getPerspectiveTransform(src, dst)
    if img_np.shape[2] == 4:
        rgb   = cv2.warpPerspective(img_np[...,:3], M, (w, h), flags=cv2.INTER_LINEAR, borderMode=cv2.BORDER_CONSTANT)
        alpha = cv2.warpPerspective(img_np[...,3],  M, (w, h), flags=cv2.INTER_LINEAR, borderMode=cv2.BORDER_CONSTANT)
        result = np.dstack([rgb, alpha])
    else:
        result = cv2.warpPerspective(img_np, M, (w, h))
    return Image.fromarray(result)

# ── AI 생성 빈 방 로드 ────────────────────────
OUTPUT_DIR = f'{DRIVE_DIR}/outputs'
LIBRARY_DIR = f'{DRIVE_DIR}/furniture_library'
target_path = f'{OUTPUT_DIR}/final_result.jpg'

if not os.path.exists(target_path):
    raise FileNotFoundError("⚠ 셀 7을 먼저 실행해서 완벽한 빈 방을 만들어주세요!")

target_img = Image.open(target_path).convert('RGB')
target_np_original = np.array(target_img)

if 'target_depth' not in dir() or target_depth.shape != target_np_original.shape[:2]:
    print('🔄 빈 방의 3D 공간 구조(Depth) 분석 중...')
    target_depth = get_depth_map(target_np_original)
    print('✅ 배경 준비 완료!')

if 'canvas_np' not in dir() or canvas_np.shape[:2] != target_np_original.shape[:2]:
    canvas_np = target_np_original.copy()

h, w = canvas_np.shape[:2]
library_files = sorted([f for f in os.listdir(LIBRARY_DIR) if f.endswith('.png')])

# ── UI 구성 ───────────────────────────────────────────────
style  = {'description_width': '150px'}
layout = widgets.Layout(width='540px')

furniture_select = widgets.Dropdown(options=library_files, description='🪑 가구 선택:', style=style, layout=widgets.Layout(width='420px'))
auto_scale_toggle = widgets.Checkbox(value=True, description='📐 깊이 기반 자동 크기', style=style)
manual_scale = widgets.FloatSlider(value=1.0, min=0.1, max=3.0, step=0.05, description='📐 크기 배율:', style=style, layout=layout, readout_format='.2f')

pos_x_slider = widgets.IntSlider(value=w//2, min=0, max=w-1, step=5, description='📍 X 좌표:', style=style, layout=layout)
pos_y_slider = widgets.IntSlider(value=h//2, min=0, max=h-1, step=5, description='📍 Y 좌표:', style=style, layout=layout)

angle_h_slider = widgets.IntSlider(value=0, min=-40, max=40, step=2, description='↔ 수평 각도(원근):', style=style, layout=layout)
angle_v_slider = widgets.IntSlider(value=0, min=-40, max=40, step=2, description='↕ 수직 각도(원근):', style=style, layout=layout)
flip_toggle = widgets.Checkbox(value=False, description='↔ 좌우 반전', style=style)
# 🔥 추가된 가이드라인 토글
show_vp_grid = widgets.Checkbox(value=True, description='📐 3D 투시 가이드라인 켜기', style=style)
preview_btn = widgets.Button(description='👁 수동 미리보기', button_style='info')
place_btn   = widgets.Button(description='✅ 배치 확정', button_style='success')
undo_btn    = widgets.Button(description='↩ 마지막 취소', button_style='warning')
reset_btn   = widgets.Button(description='🗑 캔버스 초기화', button_style='danger')
output_area = widgets.Output()

placement_history = []
placement_state   = {}

def build_furniture(furn_file, px, py, use_auto_scale, m_scale, angle_h, angle_v, flip):
    furn = Image.open(f'{LIBRARY_DIR}/{furn_file}').convert('RGBA')
    if flip: furn = furn.transpose(Image.FLIP_LEFT_RIGHT)
    if angle_h != 0 or angle_v != 0: furn = perspective_transform(furn, angle_h, angle_v)

    if use_auto_scale:
        depth_val  = get_depth_at_point(target_depth, px, py)
        auto_s     = auto_scale_by_depth(furn, depth_val, base_scale=m_scale)
        final_scale = auto_s
        depth_info  = f'깊이({depth_val:.2f}) 자동보정'
    else:
        final_scale = m_scale
        depth_info  = f'수동 크기'

    fw = max(1, int(furn.width  * final_scale))
    fh = max(1, int(furn.height * final_scale))
    return furn.resize((fw, fh), Image.LANCZOS), final_scale, depth_info

def composite(base_np, furn_pil, px, py):
    fw, fh = furn_pil.size
    x0, y0 = px - fw//2, py - fh//2
    result  = Image.fromarray(base_np).convert('RGBA')
    result.paste(furn_pil, (x0, y0), furn_pil)
    return np.array(result.convert('RGB')), (x0, y0, fw, fh)

# 🔥 핵심 수정 2: change 인자를 받아서 슬라이더 이동 시 즉각 반응하도록 변경
def on_preview(change=None):
    with output_area:
        clear_output(wait=True)
        px, py = pos_x_slider.value, pos_y_slider.value

        furn_pil, scale, depth_info = build_furniture(
            furniture_select.value, px, py, auto_scale_toggle.value, manual_scale.value,
            angle_h_slider.value, angle_v_slider.value, flip_toggle.value
        )
        preview_np, bbox = composite(canvas_np, furn_pil, px, py)

        placement_state.update({'preview': preview_np, 'bbox': bbox, 'furn_pil': furn_pil, 'file': furniture_select.value, 'pos': (px, py)})

        fig, axes = plt.subplots(1, 2, figsize=(16, 6))

        axes[0].imshow(canvas_np)
        axes[0].set_title('현재 캔버스 (가이드라인 참고)', fontsize=14)

        # 🔥 추가된 투시 그리드 로직
        if show_vp_grid.value:
            h_img, w_img = canvas_np.shape[:2]
            vp_x, vp_y = w_img // 2, h_img // 3  # 실내 사진의 일반적인 가상 소실점 위치

            # 방사형 투시선 그리기
            for x_pos in np.linspace(-w_img, w_img*2, 15):
                axes[0].plot([vp_x, x_pos], [vp_y, h_img], color='cyan', linestyle='--', alpha=0.7, linewidth=1.5)
            # 바닥 수평선 가이드
            for y_pos in np.linspace(vp_y, h_img, 6):
                axes[0].axhline(y_pos, color='magenta', linestyle=':', alpha=0.5)

        axes[1].imshow(preview_np)
        x0, y0, fw, fh = bbox
        import matplotlib.patches as patches
        axes[1].add_patch(patches.Rectangle((x0, y0), fw, fh, linewidth=2, edgecolor='lime', facecolor='none', linestyle='--'))
        axes[1].set_title(f'실시간 렌더링 | {depth_info}', fontsize=14)
        axes[1].axis('off')
        plt.tight_layout()
        plt.show()

def on_place(b):
    global canvas_np
    with output_area:
        clear_output(wait=True)
        if not placement_state: return
        placement_history.append(canvas_np.copy())
        canvas_np, _ = composite(canvas_np, placement_state['furn_pil'], placement_state['pos'][0], placement_state['pos'][1])
        print(f'✅ [{placement_state["file"]}] 안착 완료! (총 {len(placement_history)}개)')
        plt.figure(figsize=(10, 7))
        plt.imshow(canvas_np)
        plt.axis('off')
        plt.show()

def on_undo(b):
    global canvas_np
    with output_area:
        clear_output(wait=True)
        if not placement_history: return
        canvas_np = placement_history.pop()
        plt.figure(figsize=(10, 7))
        plt.imshow(canvas_np)
        plt.axis('off')
        plt.show()

def on_reset(b):
    global canvas_np
    with output_area:
        clear_output(wait=True)
        canvas_np = target_np_original.copy()
        placement_history.clear()
        placement_state.clear()
        print('🗑 캔버스 전체 초기화 완료.')

preview_btn.on_click(on_preview)
place_btn.on_click(on_place)
undo_btn.on_click(on_undo)
reset_btn.on_click(on_reset)

# 🔥 show_vp_grid 도 이벤트 리스너에 추가합니다.
for widget in [furniture_select, auto_scale_toggle, manual_scale, pos_x_slider, pos_y_slider, angle_h_slider, angle_v_slider, flip_toggle, show_vp_grid]:
    widget.observe(on_preview, names='value')

display(widgets.VBox([
    widgets.HTML('<b>🪑 가구 재배치 제어반</b>'),
    furniture_select,
    widgets.HTML('<b>📐 크기 조절</b>'),
    auto_scale_toggle, manual_scale,
    widgets.HTML('<b>📍 위치 이동</b>'),
    pos_x_slider, pos_y_slider,
    widgets.HTML('<b>🔄 3D 공간 왜곡 (원근감)</b>'),
    angle_h_slider, angle_v_slider, flip_toggle, show_vp_grid, # 🔥 여기에 추가
    widgets.HBox([preview_btn, place_btn]),
    widgets.HBox([undo_btn, reset_btn]),
    output_area,
]))

# 셀 실행 시 첫 화면 자동 렌더링
on_preview()
```

```python
# ============================================================
# 셀 10: [리얼리티 극대화] 다층 그림자 + 톤 매칭 + ControlNet 최종
# ============================================================
import cv2
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
import torch, os

# ── 상태 확인 ─────────────────────────────────────────────
if 'canvas_np' not in dir() or not placement_history:
    raise ValueError('⚠ 셀 9에서 [✅ 배치 확정]을 먼저 누르세요!')

print(f'✅ {len(placement_history)}개 배치 감지 → 리얼리티 후보정 시작')

# ── Step 1: 조명 및 색감 매칭 (톤 매칭) ───────────────────
def match_color_tone(target_np, furn_img_np, furn_mask, opacity=0.3):
    """
    배경의 평균 색감을 가구에 덧입혀 톤을 조화롭게 만듦 (따뜻함/차가움)
    """
    if furn_mask.max() == 0: return furn_img_np

    # 1. 배경의 평균 색상 추출
    avg_color = cv2.mean(target_np)[:3] # [B, G, R]

    # 2. 가구 이미지 복사
    toned_furn = furn_img_np.copy()

    # 3. 가구 마스크 영역에만 배경 평균 색상을 부드럽게 합성 (Color Overlay)
    furn_area = toned_furn[..., :3].astype(float)
    avg_color_layer = np.full_like(furn_area, avg_color)

    # 두 층을 블렌딩 (가구의 텍스처는 유지하되 색감만 입힘)
    blended = cv2.addWeighted(furn_area, 1 - opacity, avg_color_layer, opacity, 0)

    # 마스크 영역만 교체
    # furn_mask는 1채널 0 or 255. 3채널로 확장
    mask_3c = np.stack([furn_mask]*3, axis=-1) / 255.0
    toned_furn[..., :3] = (blended * mask_3c + furn_area * (1 - mask_3c)).astype(np.uint8)

    return toned_furn

# ── Step 2: 다층 그림자 생성 (스티커 느낌 탈출의 핵심) ─────
def estimate_light_direction(image_np):
    gray  = cv2.cvtColor(image_np, cv2.COLOR_RGB2GRAY).astype(float)
    h, w  = gray.shape
    left  = gray[:, :w//2].mean()
    right = gray[:, w//2:].mean()
    top   = gray[:h//2, :].mean()
    bot   = gray[h//2:, :].mean()
    light_x = (right - left) / 128.0   # 오른쪽이 밝으면 빛이 오른쪽에서 (값 양수)
    light_y = (bot - top)   / 128.0
    return np.clip(light_x, -1, 1), np.clip(light_y, -1, 1)

def generate_realistic_shadows(canvas_np, furn_mask, light_x, light_y):
    """
    두 가지 그림자를 생성하여 합성:
    1. Cast Shadow: 빛에 의해 멀리 퍼지는 부드러운 그림자 (기존)
    2. Contact Shadow: 가구 다리가 바닥에 닿는 곳에 생기는 아주 짙고 얇은 그림자 (핵심!)
    """
    h, w = furn_mask.shape
    base_shadow_mask = np.zeros((h, w), dtype=np.uint8)

    # --- 1. Cast Shadow (부드러운 그림자 - 기존 방식 개선) ---
    c_shadow_len = 0.25 # 길이
    c_shadow_opacity = 0.35 # 진하기
    c_blur_r = 31 # 블러

    # 그림자 오프셋 (조명 반대 방향)
    offset_x = int(-light_x * w * c_shadow_len)
    offset_y = int( light_y * h * c_shadow_len * 0.5)

    M = np.float32([[1, 0, offset_x], [0, 1, offset_y]])
    cast_mask = cv2.warpAffine(furn_mask, M, (w, h), borderMode=cv2.BORDER_CONSTANT, borderValue=0)
    cast_mask = cv2.GaussianBlur(cast_mask, (c_blur_r, c_blur_r), 0)

    # 가구 본체 제외
    cast_mask = cv2.subtract(cast_mask, furn_mask)


    # --- 2. Contact Shadow (짙은 접촉 그림자 - 스티커 탈출 핵심) ---
    ct_blur_r = 5 # 아주 조금만 블러
    ct_shadow_opacity = 0.8 # 아주 짙게

    # 가구 마스크를 아주 살짝만 아래로 이동 (다리 밑에만 깔리게)
    M_ct = np.float32([[1, 0, 0], [0, 1, 3]]) # 아래로 3px만
    contact_mask = cv2.warpAffine(furn_mask, M_ct, (w, h), borderMode=cv2.BORDER_CONSTANT, borderValue=0)

    # 원래 마스크 영역은 빼버림 -> 다리 밑에만 남음
    contact_mask = cv2.subtract(contact_mask, furn_mask)

    # 아주 짙고 얇게 블러 처리
    contact_mask = cv2.GaussianBlur(contact_mask, (ct_blur_r, ct_blur_r), 0)


    # --- 3. 두 그림자 합성 ---
    result = canvas_np.astype(float).copy()

    # Cast Shadow 적용
    c_alpha = (cast_mask / 255.0) * c_shadow_opacity
    for c in range(3):
        result[..., c] = result[..., c] * (1 - c_alpha)

    # Contact Shadow 적용 (위에 덧씌움)
    ct_alpha = (contact_mask / 255.0) * ct_shadow_opacity
    for c in range(3):
        result[..., c] = result[..., c] * (1 - ct_alpha)

    return np.clip(result, 0, 255).astype(np.uint8)

# ── 메인 처리 루프 ─────────────────────────────────────────────

# 1. 변경 영역 마스크 추출 (가구 전체)
diff     = np.abs(canvas_np.astype(int) - target_np_original.astype(int))
furn_mask = ((diff.sum(axis=2) > 15) * 255).astype(np.uint8)

# 노이즈 제거 (소파 쿠션 구멍 채우기 등) 어제 수정한 핵심 로직 유지
close_kernel = np.ones((25, 25), np.uint8)
furn_mask = cv2.morphologyEx(furn_mask, cv2.MORPH_CLOSE, close_kernel)
contours, _ = cv2.findContours(furn_mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
cv2.drawContours(furn_mask, contours, -1, 255, thickness=cv2.FILLED)


# 🔥 [Step 1] 글로벌 톤 매칭 적용 (가구를 배경 색감에 맞춤)
print('🎨 배경과 가구 색감 맞추는 중 (톤 매칭)...')
# canvas_np에서 가구 영역만 추출해서 톤 매칭 후 다시 덮어씀
furn_img_rgba = canvas_np.copy()
toned_furn_np = match_color_tone(target_np_original, furn_img_rgba, furn_mask, opacity=0.35)
# 톤 매칭된 가구를 원본 캔버스에 업데이트
canvas_np = toned_furn_np


# 🔥 [Step 2] 리얼 다층 그림자 생성 (바닥에 딱 붙이는 핵심 작업)
light_x, light_y = estimate_light_direction(target_np_original)
print(f'💡 추정된 조명 방향: x={light_x:.2f}, y={light_y:.2f}')

print('🌑 리얼 접촉 그림자(Contact Shadow) 생성 중...')
shadow_result = generate_realistic_shadows(
    canvas_np, furn_mask,
    light_x, light_y
)
print('✅ 그림자 생성 완료!')


# 🔥 이후의 Poisson Blending 및 ControlNet은 어제 다리가 사라지는 버그를 수정한 최신 코드를 그대로 유지합니다.
# ── Step 3: Poisson Blending ──────────────────────────────
def poisson_blend(target_np, canvas_np, mask):
    kernel  = np.ones((5, 5), np.uint8)
    mask_e  = cv2.erode(mask, kernel, iterations=2)
    if mask_e.max() == 0: return canvas_np.copy()

    contours, _ = cv2.findContours(mask_e, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    if not contours: return canvas_np.copy()

    blended = canvas_np.copy()
    for cnt in contours:
        x, y, cw, ch = cv2.boundingRect(cnt)
        region_mask   = np.zeros_like(mask_e)
        cv2.drawContours(region_mask, [cnt], -1, 255, -1)
        region_mask = cv2.erode(region_mask, kernel, iterations=1)
        if region_mask.max() == 0: continue
        cx, cy = x + cw//2, y + ch//2
        try:
            blended = cv2.seamlessClone(canvas_np, blended, region_mask, (cx, cy), cv2.NORMAL_CLONE)
        except cv2.error: continue
    return blended

print('🎨 Poisson Blending 중...')
poisson_result = poisson_blend(target_np_original, shadow_result, furn_mask)
print('✅ Poisson Blending 완료!')


# ── Step 4: ControlNet + SD 최종 보정 ────────────────────
if 'controlnet_pipe' not in dir():
    print('🤖 ControlNet Depth 모델 로드 중...')
    from diffusers import StableDiffusionControlNetInpaintPipeline, ControlNetModel
    controlnet = ControlNetModel.from_pretrained('lllyasviel/control_v11f1p_sd15_depth', torch_dtype=torch.float16)
    controlnet_pipe = StableDiffusionControlNetInpaintPipeline.from_pretrained(
        'runwayml/stable-diffusion-inpainting',
        controlnet=controlnet,
        torch_dtype=torch.float16,
        safety_checker=None,
    ).to('cuda')
    print('✅ ControlNet 로드 완료!')

# 깊이 맵 추출
new_depth = get_depth_map(poisson_result)
depth_for_ctrl = (new_depth * 255).astype(np.uint8)
depth_pil = Image.fromarray(depth_for_ctrl).convert('RGB')

# 경계 마스크
diff     = np.abs(canvas_np.astype(int) - target_np_original.astype(int))
changed  = (diff.sum(axis=2) > 10).astype(np.uint8) * 255
kernel   = np.ones((20*2+1, 20*2+1), np.uint8)
dilated  = cv2.dilate(changed, kernel, iterations=1)
eroded   = cv2.erode(changed,  kernel, iterations=1)
boundary_mask = cv2.subtract(dilated, eroded)
boundary_mask = cv2.GaussianBlur(boundary_mask, (21, 21), 0)

# 이미지 패딩 및 리사이즈
h, w  = poisson_result.shape[:2]
pad_h, pad_w = (8 - h % 8) % 8, (8 - w % 8) % 8
src_pad   = np.pad(poisson_result, ((0,pad_h),(0,pad_w),(0,0)), mode='reflect') if pad_h or pad_w else poisson_result
mask_pad  = np.pad(boundary_mask,  ((0,pad_h),(0,pad_w)),       mode='reflect') if pad_h or pad_w else boundary_mask
depth_pad = np.pad(depth_for_ctrl, ((0,pad_h),(0,pad_w)),       mode='reflect') if pad_h or pad_w else depth_for_ctrl

image_pil = Image.fromarray(src_pad)
mask_pil  = Image.fromarray(mask_pad)
depth_pil = Image.fromarray(depth_pad).convert('RGB')

# 최종 보정 (다리 삭제 버그 수정된 프롬프트 사용)
print('🎨 ControlNet 깊이 기반 최종 보정 중...')
result = controlnet_pipe(
 prompt="photorealistic interior design, perfectamente blended furniture, realistic contact shadows, high resolution, 8k",
 negative_prompt="blurry, distorted, artifacts, unnatural boundaries, harsh lines, floating objects",
 image=image_pil, mask_image=mask_pil, control_image=depth_pil,
 num_inference_steps=50, guidance_scale=7.0,
 controlnet_conditioning_scale=0.8, strength=0.25, # 낮은 강도로 다리 유지
).images[0]

result_np = np.array(result.resize((src_pad.shape[1], src_pad.shape[0])))[:h, :w]




# ── 저장 ──────────────────────────────────────────────────
OUTPUT_DIR = f'{DRIVE_DIR}/outputs'
os.makedirs(OUTPUT_DIR, exist_ok=True)

Image.fromarray(canvas_np).save(f'{OUTPUT_DIR}/step1_raw.jpg')
Image.fromarray(shadow_result).save(f'{OUTPUT_DIR}/step2_shadow.jpg')
Image.fromarray(poisson_result).save(f'{OUTPUT_DIR}/step3_poisson.jpg')
Image.fromarray(result_np).save(f'{OUTPUT_DIR}/step4_final.jpg')
print(f'✅ 저장 완료: {OUTPUT_DIR}')

# ── 5단계 비교 시각화 ─────────────────────────────────────
fig, axes = plt.subplots(1, 5, figsize=(35, 7))
titles = ['타겟 원본', '합성 (raw)', '그림자 추가', 'Poisson 보정', 'ControlNet 최종 ✅']
imgs   = [target_np_original, canvas_np, shadow_result, poisson_result, result_np]

for ax, img, title in zip(axes, imgs, titles):
    ax.imshow(img)
    ax.set_title(title, fontsize=12)
    ax.axis('off')

plt.tight_layout()
plt.savefig(f'{OUTPUT_DIR}/final_comparison.png', dpi=150)
plt.show()
print('🎉 고품질 후보정 완료!')
```
